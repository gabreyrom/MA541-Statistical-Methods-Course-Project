## La liga 2017/18

```python
la_liga_17_18 = pd.read_json("data/la_liga_2017_18.json")
```

```python
la_liga_17_18.head()
```

```python
la_liga_17_18.shape
```

```python
for x in la_liga_17_18[["match_id"]]["match_id"].to_list():
    print(x, end=" ")
```

```python
match_id = "match_1_info"

shots = []

for match_id in la_liga_17_18["match_id"]:
    with open(f"data/events/{match_id}.json") as f:
        events = json.load(f)
        for e in events:
            if e["type"]["name"] == "Shot":
                shots.append(e)

shots_df = pd.json_normalize(shots)
```
